```python
# Imports
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from imblearn.over_sampling import SMOTE
import gradio as gr

# Load dataset
df = pd.read_csv('/content/dataset.csv')
X = df.drop(columns=['target'])
y = df['target']

# One-hot encode features
X_encoded = pd.get_dummies(X, drop_first=True)
feature_names = X_encoded.columns.tolist()

# Train/test split
X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.2, random_state=42)

# ⚖️ Apply SMOTE to balance classes
smote = SMOTE(random_state=42)
X_train_balanced, y_train_balanced = smote.fit_resample(X_train, y_train)

# Train model
model = RandomForestClassifier(random_state=42)
model.fit(X_train_balanced, y_train_balanced)

# 💬 Prediction function with adjusted threshold
def predict_heart_disease(sex, age, cp, trestbps, chol, fbs, restecg, thalach,
                          exang, oldpeak, slope, ca, thal):
    sex_val = 1 if sex == "Male" else 0
    input_data = {
        'sex': sex_val,
        'age': age,
        'cp': cp,
        'trestbps': trestbps,
        'chol': chol,
        'fbs': fbs,
        'restecg': restecg,
        'thalach': thalach,
        'exang': exang,
        'oldpeak': oldpeak,
        'ca': ca,
        'thal': thal
    }
    input_df = pd.DataFrame([input_data])
    input_encoded = pd.get_dummies(input_df)

    # Add missing columns
    for col in feature_names:
        if col not in input_encoded.columns:
            input_encoded[col] = 0
    input_encoded = input_encoded[feature_names]

    # Predict with adjusted threshold
    proba = model.predict_proba(input_encoded)[0][1]
    threshold = 0.7  # 👈 You can tweak this (0.5 = default)
    prediction = 1 if proba > threshold else 0

    return f"{'🔴 HIGH RISK' if prediction == 1 else '🟢 LOW RISK'} (Risk Probability: {proba:.2f})"

# Gradio Interface
inputs = [
    gr.Dropdown(["Male", "Female"], label="Sex"),
    gr.Number(label="Age"),
    gr.Number(label="Chest Pain Type (cp)"),
    gr.Number(label="Resting Blood Pressure (trestbps)"),
    gr.Number(label="Cholesterol (chol)"),
    gr.Number(label="Fasting Blood Sugar >120 (fbs)"),
    gr.Number(label="Resting ECG (restecg)"),
    gr.Number(label="Max Heart Rate (thalach)"),
    gr.Number(label="Exercise-induced(exr)"),
    gr.Number(label="ST Depression (oldpeak)"),
    gr.Number(label="Number of Vessels (ca)"),
    gr.Number(label="Thalassemia (thal)")
]

gr.Interface(
    fn=predict_heart_disease,
    inputs=inputs,
    outputs="text",
    title="❤️ Heart Disease Prediction ",
    description="This Is your heart's best friend.",
    theme="default"
).launch(share=True)
```

```python
!pip install gradio
!pip install imbalanced-learn
!pip install imbalanced-learn
import gradio as gr
import pandas as pd
import joblib
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split, GridSearchCV
from imblearn.over_sampling import SMOTE

# ✅ Load and prepare dataset
df = pd.read_csv('datasetNEW.csv')
df = df.rename(columns={'weight ': 'weight'})  # Fix column name with trailing space
df.dropna(inplace=True)

# ✅ Prepare features and target
X = df.drop("target", axis=1)
y = df["target"]

# ✅ Feature Engineering: Add BMI (using correct column names)
X['bmi'] = X['weight'] / ((X['height']/100) ** 2)

# One-hot encode categorical features
X_encoded = pd.get_dummies(X)
feature_names = X_encoded.columns.tolist()

# ✅ Train/test split with stratification
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

# ✅ SMOTE Application
smote = SMOTE(random_state=42)
X_train_balanced, y_train_balanced = smote.fit_resample(X_train, y_train)

# ✅ Hyperparameter Tuning
param_grid = {
    'n_estimators': [100, 200],
    'max_depth': [None, 10],
    'min_samples_split': [2, 5],
    'min_samples_leaf': [1, 2]
}

model = GridSearchCV(
    RandomForestClassifier(random_state=42),
    param_grid,
    cv=5,
    scoring='recall',  # Better for medical applications
    n_jobs=-1
)
model.fit(X_train_balanced, y_train_balanced)

# Save best model
best_model = model.best_estimator_
joblib.dump(best_model, "heart_model.pkl")
joblib.dump(feature_names, "features.pkl")

# 🎯 Enhanced prediction function
def predict_heart_disease(sex, age, cp, trestbps, chol, fbs, restecg, thalach,
                         exang, oldpeak, slope, ca, thal, height, weight, smoking_history):
    # Convert inputs with correct column names
    input_data = {
        'sex': 1 if sex == "Male" else 0,
        'age': age,
        'cp': cp,
        'trestbps': trestbps,
        'chol': chol,
        'fbs': fbs,
        'restecg': restecg,
        'thalach': thalach,
        'exang': exang,
        'oldpeak': oldpeak,
        'slope': slope,
        'ca': ca,
        'thal': thal,
        'height': height,
        'weight': weight,  # Match cleaned column name
        'smoking_history': 1 if smoking_history == "Yes" else 0,
        'bmi': weight / ((height/100) ** 2)
    }

    # Create DataFrame
    df_input = pd.DataFrame([input_data])

    # Align features with training set
    df_input = pd.get_dummies(df_input)
    for col in feature_names:
        if col not in df_input.columns:
            df_input[col] = 0

    df_input = df_input[feature_names]

    # Predict with optimized threshold
    proba = best_model.predict_proba(df_input)[0][1]
    prediction = 1 if proba > 0.65 else 0  # ROC-optimized threshold

    return f"""
    {'🔴 HIGH RISK' if prediction == 1 else '🟢 LOW RISK'}
    Probability: {proba:.2%}
    Key Factors: {', '.join(best_model.feature_names_in_[np.argsort(best_model.feature_importances_)[-3:]])}
    """

# 🌈 Gradio interface
with gr.Blocks(theme=gr.themes.Soft(primary_hue="pink")) as demo:
    gr.Markdown("""## ❤ Advanced Heart Disease Predictor""")

    with gr.Row():
        with gr.Column():
            sex = gr.Dropdown(["Male", "Female"], label="Gender")
            age = gr.Slider(20, 100, label="Age")
            height = gr.Number(label="Height (cm)")
            weight = gr.Number(label="Weight (kg)")
            smoking_history = gr.Radio(["Yes", "No"], label="Smoking History")

        with gr.Column():
            cp = gr.Slider(0, 3, step=1, label="Chest Pain Type")
            trestbps = gr.Number(label="Resting Blood Pressure")
            chol = gr.Number(label="Cholesterol")

    with gr.Row():
        fbs = gr.Radio([0, 1], label="Fasting Blood Sugar > 120")
        restecg = gr.Radio(["Normal", "Moderate", "Extreme"], label="Resting ECG")
        thalach = gr.Number(label="Max Heart Rate")
        exang = gr.Radio(["No", "Yes"], label="Exercise")

    with gr.Row():
        oldpeak = gr.Slider(0.0, 2.0, label="ST Depression")
        slope = gr.Slider(0, 2, step=1, label="Slope")
        ca = gr.Slider(0, 4, step=1, label="Major Vessels")
        thal = gr.Slider(0, 3, step=1, label="Thalassemia")

    submit_btn = gr.Button("Analyze Risk")
    result = gr.Markdown()

    submit_btn.click(
        predict_heart_disease,
        inputs=[sex, age, cp, trestbps, chol, fbs, restecg, thalach,
                exang, oldpeak, slope, ca, thal, height, weight, smoking_history],
        outputs=result
    )

demo.launch(share=True)
```
